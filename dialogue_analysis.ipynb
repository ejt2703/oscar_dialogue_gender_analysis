# Data Analysis for Column Charts - Gender by Individual Film (Viz 2 & 3)
## "Men do most of the talking, but women take the lead in 'Sentimental Value' and 'Hamnet'"
## "Last year, women spoke the majority of words in three films by over at least 60%"
Below, I create a series of pivot tables showing the number of words spoken by each gender per film. Then, I save them as .CSV files -- one for each award year.

```python
# Here I create a pivot table for the 98th Awards with films as rows and genders as columns.

gender_by_film_98 = awards_98th.pivot_table(
    values='word_count_corrected',
    index='film',
    columns='speaker_gender',
    aggfunc='sum',
    fill_value=0
)

# Here I add a column for total words.
gender_by_film_98['TOTAL'] = gender_by_film_98.sum(axis=1)

# Here I add a totals row for each gender.
totals_row = gender_by_film_98.sum(axis=0)
totals_row.name = 'TOTAL'
gender_by_film_98 = pd.concat([gender_by_film_98, totals_row.to_frame().T])

# And here I save it as a .CSV file.
gender_by_film_98.to_csv('gender_by_film_98.csv')
```

```python
gender_by_film_98
```

```python
# This is the same as the above, except for the 97th Awards.

gender_by_film_97 = awards_97th.pivot_table(
    values='word_count_corrected',
    index='film',
    columns='speaker_gender',
    aggfunc='sum',
    fill_value=0
)

gender_by_film_97['TOTAL'] = gender_by_film_97.sum(axis=1)

totals_row = gender_by_film_97.sum(axis=0)
totals_row.name = 'TOTAL'
gender_by_film_97 = pd.concat([gender_by_film_97, totals_row.to_frame().T])

gender_by_film_97.to_csv('gender_by_film_97.csv')
```

# Data Analysis for Bar Charts - Words per Lead Character + Their Gender (Viz 4 & 5)
## "How much do lead characters speak in their films?"
## And, "Last year featured more lead women than men -- and they generally spoke more than them too"
Below, I create pivot tables showing the total number of words spoken by each lead character (and their film and gender) for each award year, and save them as .CSV files. Afterwords, I input the data into Flourish to create the bar charts.

```python
# Here, I create a pivot table that filter out lead characters, their films, their genders, and the sum of their word counts for the 98th Awards.
lead_character_wc_98th = awards_98th[awards_98th['speaker_role'] == 'LEAD'].groupby(['film', 'speaker']).agg({
    'word_count_corrected': 'sum',
    'speaker_gender': 'first'
}).reset_index()

# Here, I add a column that shows the total word count for each film, subtracted by the lead character's word count.
# In Flourish, this allows me to show the lead character's word count in the context of the rest of the film's dialogue.
film_total_wc = awards_98th.groupby('film')['word_count_corrected'].sum()
lead_character_wc_98th['other_characters_wc'] = lead_character_wc_98th['film'].map(film_total_wc) - lead_character_wc_98th['word_count_corrected']

lead_character_wc_98th.to_csv('lead_character_wc_98th.csv', index=False)

lead_character_wc_98th
```

```python
# And here is the same as above for the 97th Awards.
lead_character_wc_97th = awards_97th[awards_97th['speaker_role'] == 'LEAD'].groupby(['film', 'speaker']).agg({
    'word_count_corrected': 'sum',
    'speaker_gender': 'first'
}).reset_index()

film_total_wc = awards_97th.groupby('film')['word_count_corrected'].sum()
lead_character_wc_97th['other_characters_wc'] = lead_character_wc_97th['film'].map(film_total_wc) - lead_character_wc_97th['word_count_corrected']

lead_character_wc_97th.to_csv('lead_character_wc_97th.csv', index=False)

lead_character_wc_97th
```

# Data Analysis for Heatmap (Viz 6)
## "Who do characters talk about more: themselves or others?"

I wanted to see if I could measure characters' interest in themselves vs. others, and compare that across genders. So, I analyzed the number of select "I" words and "you" words for each gender per movie, and was able to determine a ratio that shows the degree to which characters speak more or less self-referentially.

This segment code was created with assistance from Claude AI, and was manually fact-checked with a random sample of movies.

The code creates a function that counts the number of words characters use that refer to themselves ("I" words) and the number of words used that refer to the people they speak to ("you" words). Then, the code compares them as a "I-to-You" for each gender's usage per film.

The analysis that comes from this code has some limitations. It excludes first-person plural pronouns like "we" and "us". Also, if characters refer to someone else by their name (and not by a "you"), it isn't counted, which means some words (names) characters say referring to others aren't counted. It also doesn't count third person pronouns, like "he," "she" or "they."

While this code might be useful for getting an idea of characters' interest in themselves vs. others, it of course has its limitations and should not be used as a definitive evaluation.

```python
# This defines the main function
def analyze_words(source, label=None):  # 'source' is the data, 'label' is an optional title like a film name
    """
    Count I-to-YOU pronoun ratio by gender.

    Parameters:
        source  : pd.DataFrame or str path to a CSV file
        label   : optional label for the output header (e.g. film name)
    """
    # ^ this is a docstring: a plain-English description of the function, not executed as code

    # This defines the words I'm looking for: I words and you words.
    # Each list is created twice — once with straight apostrophes and once with curly apostrophes —
    # so the code catches contractions regardless of how they were typed in the source data.
    curly = chr(8217)  # stores the curly apostrophe character ' (as opposed to the straight one ') so contractions like "i'm" can be matched in both styles

    you_words = {'you', 'your', 'yours', 'yourself', "you're", "you've", "you'll", "you'd", "y'all", 'you-', 'your-', 'yours-', 'yourself-', "you're-", "you've-", "you'll-", "you'd-", "y'all-", 'yourselves', 'yourselves-', 'thou', 'thou-', 'thy', 'thy-', 'thee', 'thee-', 'thine', 'thine-'}  # defines a set of all "you"-related words to look for
    you_words.update([w.replace("'", curly) for w in you_words])  # adds curly-apostrophe versions of each contraction to the same set, so both styles are covered

    i_words = {'i', "i'm", "i've", "i'll", "i'd", 'my', 'mine', 'me', 'myself', 'i-', "i'm-", "i've-", "i'll-", "i'd-", 'my-', 'mine-', 'me-', 'myself-'}  # defines a set of all "I"-related words to look for
    i_words.update([w.replace("'", curly) for w in i_words])  # adds curly-apostrophe versions of each contraction, same as above

    # This loads the data from the source.
    if isinstance(source, str):  # checks whether 'source' is a file path (a string) rather than a ready-made table of data
        df = pd.read_csv(source)  # if so, reads the CSV file into a DataFrame (a table) called 'df'
        label = label or source  # if no label was provided, uses the file path as the label
    else:
        df = source  # if 'source' is already a DataFrame, just use it directly

    # This checks that the data has the two columns this code needs to work: speaker_gender and dialogue.
    # If not, it stops immediately and explains what's missing.
    required_cols = {'speaker_gender', 'dialogue'}
    if not required_cols.issubset(df.columns):  # checks whether both required columns exist in the data
        raise ValueError(f"Data must contain columns: {required_cols}. Found: {set(df.columns)}")  # if not, stops the code and prints a helpful error message

    # This finds the genders in the data.
    # Looks at the data and automatically detects however many genders are present,
    # so the code works no matter what the data contains.
    genders = sorted(df['speaker_gender'].dropna().unique().tolist())  # finds all unique gender values in the data, removes blanks, and sorts them alphabetically into a list

    # This is a smaller function used inside the main one.
    # It takes a set of dialogue lines and counts the "I"-words and "you"-words within them.
    def count_pronouns(dialogues):
        all_text = ' '.join(dialogues.astype(str))  # joins all dialogue lines into one long string with spaces between them
        words = re.findall(r"\b[A-Z]+(?:['\u2019][A-Z]+)*\b", all_text)  # extracts every word (including contractions) from the text using a pattern; \u2019 is the curly apostrophe, \b marks word boundaries - does not extract words where "I" is part of the word
        words = [w.lower() for w in words]  # converts every word to lowercase so "I" and "i" are treated the same

        you_count = 0  # sets a counter for "you"-words, starting at zero
        i_count = 0  # sets a counter for "I"-words, starting at zero

        for word in words:  # This loops through every word one at a time
            if word in you_words:  # checks if the word is in the "you" set
                you_count += 1  # if so, adds 1 to the "you" counter
            # If the word matched you, you should not also count it as an "I" word, 
            # so I would continue to the next.
            # Or you can keep the below and raise an error. That would be something I'd want to 
            # know b/c it's unexpected and could be an indication of miscounts
            if word in i_words:  # checks if the word is in the "I" set (separate check, not else/if)
                i_count += 1  # if so, adds 1 to the "I" counter

        return you_count, i_count  # sends both counts back to wherever the function was called from

    # This counts each word per gender.
    # Splits the data by gender, runs the word-counting function on each group,
    # and stores the results ready to be printed.
    you_counts = {}  # creates an empty dictionary to store "you"-word counts per gender
    i_counts = {}  # creates an empty dictionary to store "I"-word counts per gender
    for gender in genders:  # loops through each gender found in the data
        dialogues = df[df['speaker_gender'] == gender]['dialogue']  # filters the data to just that gender's rows and grabs the dialogue column
        you_counts[gender], i_counts[gender] = count_pronouns(dialogues)  # runs count_pronouns on those dialogues and stores the two results

    # This prints the table header.
    # Prints the film name (or label) and the column headings at the top of the output table.
    header = f" {label} " if label else ""  # wraps the label in spaces if one was provided, otherwise leaves it blank
    print(f"\n{'='*10}{header}{'='*10}")  # prints a header line like "========== Hamnet =========="

    header_row = f"{'':20}"  # starts the column header row with a blank 20-character space for the label column
    for gender in genders:  # loops through each gender
        header_row += f"  {gender:>8}"  # adds the gender name right-aligned in an 8-character field
    print(header_row)  # prints the completed column header row
    print("-" * (20 + len(genders) * 10))  # prints a dividing line sized to match the table width

    # This prints the raw number of "I"-words and "you"-words each gender uses.
    i_row = f"{'  I words':<20}"  # starts the "I words" row with its label
    for gender in genders:  # loops through each gender
        i_row += f"  {i_counts[gender]:>8}"  # adds that gender's "I"-word count
    print(i_row)  # prints the completed "I words" row

    you_row = f"{'  YOU words':<20}"  # starts the "YOU words" row with its label
    for gender in genders:  # loops through each gender
        you_row += f"  {you_counts[gender]:>8}"  # adds that gender's "you"-word count
    print(you_row)  # prints the completed "YOU words" row

    print("-" * (20 + len(genders) * 10))  # prints a final dividing line before the ratio

    # This prints the I-to-YOU ratio.
    # It divides each gender's "I"-word count by their "you"-word count to produce a ratio.
    # A ratio above 1 means they use more "I"-words than "you"-words, and vice versa.
    ratio_row = f"{'  I-to-YOU ratio':<20}"  # starts the ratio row with its label
    for gender in genders:  # loops through each gender
        i = i_counts[gender]  # retrieves that gender's "I"-word count
        y = you_counts[gender]  # retrieves that gender's "you"-word count
        ratio = (i / y) if y > 0 else 0  # divides I by YOU to get the ratio; uses 0 if there are no "you"-words to avoid a division-by-zero error
        ratio_row += f"  {ratio:>7.2f}x"  # adds the ratio formatted to 2 decimal places with an "x" suffix
    print(ratio_row)  # prints the completed ratio row
```

```python
# Here I apply the analyze_words function to the 98th Award films, which allows me to analyze their Self-to-Other ratios by gender.

analyze_words(hamnet, label='HAMNET')
analyze_words(sentimental_value, label='SENTIMENTAL VALUE')
analyze_words(sinners, label='SINNERS')
analyze_words(bugonia, label='BUGONIA')
analyze_words(train_dreams, label='TRAIN DREAMS')
analyze_words(marty_supreme, label='MARTY SUPREME')
analyze_words(f_one, label='F-ONE')
analyze_words(frankenstein, label='FRANKENSTEIN')
analyze_words(one_battle_after_another, label='ONE BATTLE AFTER ANOTHER')
analyze_words(the_secret_agent, label='THE SECRET AGENT')
```
